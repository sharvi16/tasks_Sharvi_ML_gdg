Now, let's some employ some Feature Engineering tactics which you would have performed in the previous tasks.

Create a function called crab_processing which takes a dataframe as an input, performs the following tasks, and returns a cleaned and edited df:

1) Drop the 'id' column.

2) Drop all rows where 'Height' of the crab is 0.

3) Create a new column called 'LostWeight' which is calculated using the following formula - ```
Weight - (Shucked + Viscera + Shell)``` .
If this formula gives a number less than 0, then assign 0 to the LostWeight column, else assign 1.

4) Finally, use one hot encoding to encode the column 'Sex'

Take help from your previous task where you have implemented these steps in different cells.


````python
import numpy as np

def crab_processing(df):
  df.drop('id',axis=1,inplace=True)
  df=df[df['Height']>0]
  df['LostWeight'] = df['Weight'] - (df['Shucked Weight'] + df['Viscera Weight'] + df['Shell Weight'])
  df.loc[df['LostWeight'] < 0, 'LostWeight'] = 0
  df.loc[df['LostWeight'] >= 0, 'LostWeight'] = 1

  df=pd.get_dummies(df,columns=['Sex'])
  return df
````

### Choosing optimal 'k'

**Above, we chose n_neighbours to be equal to 1. Choosing a small value of K leads to unstable decision boundaries. <br>
We need to select n_neighbours by calculating the accuracy for every value of n from 1 to 60 and then choosing the one which gives the highest accuracy.**

````python
# Do the same as we did above, but this time make a loop from n = 1 to n = 60 and append the accuracy
# for each in a list

# Your code starts here.
accuracy_rate = []

for i in range(1, 61):
    knn = KNeighborsClassifier(n_neighbors=i)
    knn.fit(X_train, y_train)
    pred_i = knn.predict(X_test)
    accuracy_rate.append(np.mean(pred_i == y_test))
# Your code ends here
````

**Plot a graph of K value vs Accuracy**

````python
# Your code starts here.
plt.figure(figsize=(10, 6))
plt.plot(range(1, 61), accuracy_rate, color='blue', linestyle='dashed', marker='o', markerfacecolor='green', markersize=10)
plt.title('Accuracy Rate vs. K Value')
plt.xlabel('K')
plt.ylabel('Accuracy Rate')
plt.grid(True)
plt.show()
# Your code ends here
````

**Choose the best value of n_neighbours and give a reason why and also print the accuracy**

````python
# Your code starts here.
import numpy as np

optimal_k_index = np.argmax(accuracy_rate)
optimal_k = optimal_k_index + 1
max_accuracy = accuracy_rate[optimal_k_index]

print(f"{optimal_k}")
print(f"{max_accuracy:.3f}")
# Your code ends here
````

## Reinforcement Learning

Reinforcement Learning (RL) is a branch of machine learning where an agent learns by doing things and seeing what happens next. Unlike supervised learning, which relies on labeled data, or unsupervised learning, which finds patterns in unlabeled data, RL explores, make mistakes, and learn from rewards and penalties.

It’s a lot like training a dog. You don’t give it step-by-step instructions.
You simply reward it when it does the right thing and avoid rewarding it when it does something wrong. Over time, the dog understands what behavior gets treats.
RL works the same way: the agent takes an action, the environment reacts, and the agent receives a reward. The agent’s goal is to learn which actions will give it the most reward in the long run.

### Key Components of Reinforcement Learning:

*   **Agent**: The learner or decision-maker.
*   **Environment**: The world with which the agent interacts.
*   **State**: The current situation of the agent in the environment.
*   **Action**: What the agent can do in a given state.
*   **Reward**: A score or feedback that tells the agent how good or bad its action was.
*   **Policy**: The agent's strategy, mapping states to actions.

### How it Works:

1.  **Observation**: The agent observes its current state in the environment.
2.  **Action Selection**: Based on its current policy, the agent chooses an action.
3.  **Execution**: The agent performs the chosen action.
4.  **Reward and New State**: The environment transitions to a new state and provides a reward signal to the agent.
5.  **Learning**: The agent updates its policy based on the reward received, aiming to choose actions that lead to higher rewards in the future.

### Where is RL Used?

*   **Autonomous Driving**: Developing decision-making systems for self-driving cars.
*   **Games**: Systems like AlphaGo and DeepMind’s agents have beaten world champions and mastered complex video games.

### Challenges and the Future:

Even though RL is powerful, it isn’t perfect. It usually needs a lot of practice, which often means millions of trials in simulations. The agent also needs to balance exploring new actions vs using actions it already knows work well. And in real-world situations, making mistakes can be risky.
However, with ongoing research and advancements, RL continues to push the boundaries of what machines can learn and achieve, paving the way for more intelligent and adaptive AI systems.

## Oversampling and Undersampling:

In machine learning, one common problem is working with imbalanced datasets. This means one class has many more examples than the other. For example, in fraud detection, real transactions are everywhere, while fraud ones are rare.

If we train a model on such imbalanced data, it may end up predicting only the majority class because that alone gives high accuracy but completely fails at detecting the minority class, which is usually the one we are predicting.

To fix this, we use techniques like oversampling and undersampling to rebalance the dataset.

### Undersampling: Reducing the Majority

**What it is:** Undersampling involves reducing the number of samples in the majority class to match the number of samples in the minority class.

**How it works:**
* The simplest way is randomly delete some majority class samples.
* More advanced methods (like NearMiss) carefully pick which majority samples to keep based on how close they are to minority samples so we don’t lose important information.

**When to use it:** Undersampling is useful when you have a very large dataset, and reducing its size won't lead to a significant loss of information. It can help reduce training time and computational cost.

**Potential drawbacks:** The biggest issue is that you might accidentally remove useful data, which can make the model less accurate.

### Oversampling: Increasing the Minority

**What it is:** Oversampling involves increasing the number of samples in the minority class to match the number of samples in the majority class.

**How it works:**
*   **Random Oversampling:** Simply duplicates random instances from the minority class.
*   **SMOTE (Synthetic Minority Over-sampling Technique):** This is a more advanced method. Instead of just duplicating existing minority samples, SMOTE creates synthetic new samples that are similar to existing minority samples but not exact copies. It does this by taking a minority sample, finding its k-nearest neighbors, and creating new samples along the line segments connecting the original sample to its neighbors.

**When to use it:** Oversampling is generally preferred when undersampling would lead to a significant loss of valuable data, or when the dataset is not extremely large.

**Potential drawbacks:** Random oversampling can lead to overfitting, as it just duplicates existing samples. SMOTE mitigates this by creating synthetic samples, but it can sometimes create noisy samples if the minority class is very sparse.

### Which to Choose?

The best approach often depends on your specific dataset and problem. Sometimes, a combination of both oversampling and undersampling (hybrid methods) yields the best results. It's crucial to experiment with different techniques and evaluate your model's performance using appropriate metrics that are sensitive to class imbalance, rather than just accuracy.
